```python
class ItemFactura:
    def __init__(self, nombre: str, precio: float, cantidad: int):
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad

    def calcular_subtotal(self) -> float:
        return self.precio * self.cantidad


class Factura:
    def __init__(self):
        self.items = []

    def agregar_item(self, nombre: str, precio: float, cantidad: int):
        nuevo_item = ItemFactura(nombre, precio, cantidad)
        self.items.append(nuevo_item)

    def calcular_total(self) -> float:
        total = 0.0
        for item in self.items:
            total += item.calcular_subtotal()
        return total

    def mostrar_factura(self):
        print("=" * 45)
        print("               DETALLE DE FACTURA            ")
        print("=" * 45)
        print(f"{'Producto':<20} | {'Cant.':<5} | {'Precio':<8} | {'Subtotal':<8}")
        print("-" * 45)
        
        for item in self.items:
            subtotal = item.calcular_subtotal()
            print(f"{item.nombre:<20} | {item.cantidad:<5} | ${item.precio:<7.0f} | ${subtotal:<7.0f}")
            
        print("-" * 45)
        print(f"TOTAL A PAGAR: ${self.calcular_total():.0f}")
        print("=" * 45)


if __name__ == "__main__":
    mi_factura = Factura()

    mi_factura.agregar_item("Coquita", 25000, 2)
    mi_factura.agregar_item("Malboro", 45000, 1)
    mi_factura.agregar_item("Alfombrilla", 8000, 3)

    mi_factura.mostrar_factura()
```
