```python
class Guerrero:
    def atacar(self):
        return "¡   Pichulazo crítico!"

class Mago:
    def atacar(self):
        return "¡Lanza una bola de fuego!"

class Arquero:
    def atacar(self):
        return "¡Dispara una flecha certera!"

def realizar_combate(personajes):
    for p in personajes:
        print(p.atacar())

if __name__ == "__main__":
    lista_personajes = [Guerrero(), Mago(), Arquero()]
    
    realizar_combate(lista_personajes)
```
