```python
class Arma:
    def __init__(self, nombre : str, danio : int):
        self.nombre = nombre
        self.danio = danio

    def atacar(self) -> str:
        return f"{self.nombre} causa {self.danio} puntos de daño."

class Guerrero:
    def __init__ (self, nombre : str, puntos_vida : int):
        self.nombre = nombre
        self.puntos_vida = puntos_vida
        self.ses_arma = None

    def equipar_arma(self, arma: Arma):
        self.ses_arma = arma

    def realizar_ataque(self):
            if self.ses_arma:
                print(f"{self.nombre} ataca: {self.ses_arma.atacar()}")
            else:
                print(f"{self.nombre} ataca con los puños (1 punto de daño).")

guerrero1 = Guerrero("Aragorn", 100)
espada = Arma("Espada de Acero", 30)

guerrero1.realizar_ataque()

guerrero1.equipar_arma(espada)
guerrero1.realizar_ataque()
```

```python
class Mago:
    def __init__(self, nombre: str, puntos_vida: int, mana: int):
        self.nombre = nombre
        self.puntos_vida = puntos_vida
        self.mana = mana
        self.hechizo = None

    def equipar_hechizo(self, hechizo: str):
        self.hechizo = hechizo

    def realizar_ataque(self):
        if self.hechizo:
            print(f"{self.nombre} lanza {self.hechizo}!")
        else:
            print(f"{self.nombre} ataca con su bastón (1 punto de daño).")

mago1 = Mago("Gandalf", 80, 100)
mago1.realizar_ataque()

mago1.equipar_hechizo("Bola de Fuego")
mago1.realizar_ataque()
```
