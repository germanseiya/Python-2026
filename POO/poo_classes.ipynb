```python
class Dog:
    """Un intento sencillo de modelar un perro."""

    def __init__(self, name, age):
        """Inicializa los atributos de nombre y edad."""
        self.name = name
        self.age = age

    def sit(self):
        """Simula un perro sentándose en respuesta a una orden."""
        print(f"{self.name} se sienta.")

    def rueda(self):
        """Simula hacer la croqueta en respuesta a una orden."""
        print(f"{self.name} está haciendo la croqueta!")


    def saltar(self):
        """Simula hacer un salto en respuesta a una orden."""
        print(f"{self.name} está saltando!")

    def ladrar(self):
        """Simula hacer un ladrido respuesta a una orden."""
        print(f"{self.name} Está ladrando!")

    def correr(self):
        """Simula correr en respuesta a una orden."""
        print(f"{self.name} está corriendo!")

    def detenerse(self):
        """simula detenerse en respuesta a una orden."""
        print(f"{self.name} se detiene!")

my_dog = Dog('Ichika', 6)

print(f"Mi perra se llama {my_dog.name}.")
print(f"Mi perra tiene {my_dog.age} años.")

my_dog.sit()
my_dog.rueda()

your_dog = Dog('Makima', 3)
print(f"\nMi otro perra se llama {your_dog.name}.")
print(f"Mi otro perra tiene {your_dog.age} años.")
your_dog.sit()
your_dog.rueda()

my_dog.correr()
my_dog.saltar()
my_dog.correr()
my_dog.detenerse()
```

```python
class Alcancia:
    def __init__(self):
        self.saldo = 0.0 

    def guardar_dinero(self, cantidad):
        if cantidad > 0:
            self.saldo = self.saldo + cantidad
            print(f"¡Has guardado ${cantidad}! Saldo actual: ${self.saldo}")
        else:
            print("Error: No puedes guardar cantidades negativas o vacías.")

    # MÉTODO para consultar el estado actual
    def ver_saldo(self):
        print(f"Saldo total acumulado: ${self.saldo}")

mi_cucha = Alcancia()

mi_cucha.ver_saldo()
mi_cucha.guardar_dinero(50)
mi_cucha.guardar_dinero(20.5)
mi_cucha.ver_saldo()
```

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor  

    def obtener_informacion(self):
        return f"'{self.titulo}', escrito por {self.autor}"


libro_favorito = Libro("Mein Kampf", "Adolf Hitler")
libro_estudio = Libro("Curso Intensivo de Python", "Eric Matthes")

info_uno = libro_favorito.obtener_informacion()
info_dos = libro_estudio.obtener_informacion()

print("Fichas bibliográficas generadas:")
print(info_uno)
print(info_dos)
```
