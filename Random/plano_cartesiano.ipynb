```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
eje_x = [3, 1, 0, -1, -3]
eje_y = [10, 2, 1, 2, 10]

plt.figure(figsize=(8, 4))
plt.plot(eje_x, eje_y, color='green', marker='o', linestyle='--', linewidth=2)

plt.title('Parabola', fontsize=14)
plt.xlabel('Eje X', fontsize=10)
plt.ylabel('Eje Y', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.8) 

plt.show()
```

```python
eje_x = [2, 1, 0, -1, -2]
eje_y = [1, -1, -3, -5, -7]

fig, ax = plt.subplots(figsize=(6, 6))
limite = 10
ax.set_xlim(-limite, limite)
ax.set_ylim(-limite, limite)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')
ax.set_xticks(np.arange(-limite, limite + 1, 1))
ax.set_yticks(np.arange(-limite, limite + 1, 1))

ax.set_aspect('equal')
plt.plot(eje_x, eje_y, color='green', linestyle='-', marker='o', markersize=6, zorder=5)
plt.show()
```
