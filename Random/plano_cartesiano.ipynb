```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
eje_x = [3, 1, 0, -1, -3]
eje_y = [10, 2, 1, 2, 10]

fig, ax = plt.subplots(figsize=(6, 6))
limite = 10
ax.set_xlim(-limite, limite)
ax.set_ylim(-limite, limite)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')
ax.set_xticks(np.arange(-limite, limite + 1, 1))
ax.set_yticks(np.arange(-limite, limite + 1, 1))

ax.set_aspect('equal')
plt.plot(eje_x, eje_y, color='green', linestyle='-', marker='o', markersize=6, zorder=5)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
limite = 10
ax.set_xlim(-limite, limite)
ax.set_ylim(-limite, limite)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')
ax.set_xticks(np.arange(-limite, limite + 1, 1))
ax.set_yticks(np.arange(-limite, limite + 1, 1))
ax.set_aspect('equal')

h, k = 2, 3
radio = 2

circulo = plt.Circle((h, k), radio, color='green', fill=False, linewidth=2, label='$(x-2)^2 + (y-3)^2 = 4$')
ax.add_patch(circulo)

plt.scatter(h, k, color='black', zorder=5, label='Centro C(2,3)')

plt.legend(loc='upper right')
plt.show()
```
